```python
from typing import Any


class LoanPrediction:
    def __init__(self, in_features, out_features) -> None:
        #* Linear layers
        self.layer1 = Linear(in_features= in_features, out_features= 30)
        self.layer2 = Linear(in_features= 30, out_features= 30)
        self.layer3 = Linear(in_features= 30, out_features= out_features)
        
        #* activations
        self.act = GELU()
    
    def forward(self, inputs):
        out = self.layer3(self.act(self.layer2(self.act(self.layer1(inputs)))))
        return out
    
    def __call__(self, inputs) -> Any:
        return self.forward(inputs)
```
